### Selección de Supervivientes (Steady-State)

Les sugiero una variante clásica en lugar de la variante que eligieron.
Esta variante consiste en reemplazar a las n peores soluciones de la población
actual por las n mejores soluciones de las soluciones obtenidas luego de la 
secuencia selección de padres, cruce, mutación y BL.
Les hago esta sugerencia porque, al usar esta variante, los algoritmos suelen funcionar
de manera más eficiente.

- Probabilidad de cruce y probabilidad de mutación
Deberían considerar dos valores para cada una de ellas, para luego
poder evaluar el algoritmo con distintas configuraciones.
- Tamaño de la población
También deberían considerar dos valores posibles para el tamaño de la población.
Actualmente, es viable considerar un valor de 100 para el tamaño de la población.
- Condición de finalización:
Les sugiero considerar solamente el hecho de alcanzar un número máximo de 
generaciones (1000 o más).
Esta sugerencia se debe a que definir el criterio de convergencia (número de generaciones
sin mejora) de forma acertada suele ser difícil cuando no se tiene experiencia previa en el 
uso de este tipo de algoritmos. Si el criterio no es definido de forma acertada, se puede
detener el algoritmo prematuramente, y evitar que alcance la mejor solución que podría alcanzar.
